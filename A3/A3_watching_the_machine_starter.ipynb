```python
import pandas as pd

responses = [
    "Yes", "yes", "No", "YES", "No",
    "yes", "", "No", "Yes", "no",
    "YES", "Yes", "", "No", "yes",
    "No", "YES", "no", "Yes", ""
]

survey = pd.DataFrame({
    "participant_id": range(1, 21),
    "response": responses
})

survey.to_csv("a3_survey_data.csv", index=False)

display(survey)
```

```python
cleaned = survey["response"].str.strip().str.lower()
cleaned = cleaned.replace("", "missing")

counts = cleaned.value_counts()
percentages = (counts / len(survey) * 100).round(2)

results = pd.DataFrame({
    "Count": counts,
    "Percentage (%)": percentages
})

display(results)
print("Total responses:", len(survey))
print("Percentage sum:", percentages.sum())
```
