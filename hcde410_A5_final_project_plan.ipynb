My research question is: **What types of content do sports betting companies use on YouTube, and how do they incorporate promotional strategies into sports-related and entertainment content?**

I plan to compare DraftKings and FanDuel to understand how frequently they use promotional offers, sports entertainment, and direct calls to action.

I plan to study publicly available videos from the official YouTube channels of DraftKings and FanDuel. My primary data source will be YouTube video metadata, accessed through the YouTube Data API (https://developers.google.com/youtube/v3). I plan to examine approximately 50–100 videos in total, ideally 25–50 from each company, within a consistent time period. I will identify a dominant video format and separately code the presence of promotional offers, sports-related entertainment, direct calls to action, and promo codes, allowing multiple indicators to be true for one video. I will then examine how frequently these strategies appear and how they differ between the two companies. If permitted by the API terms, I may also explore engagement statistics as a secondary part of the analysis.

The data is subject to the YouTube API Services Terms of Service (https://developers.google.com/youtube/terms/api-services-terms-of-service) and Developer Policies (https://developers.google.com/youtube/terms/developer-policies). I will verify the permitted collection, analysis, and storage methods before beginning. Since gambling advertising can influence vulnerable audiences, I also want to consider the ethical implications of these promotional strategies. ## A5 — Overview & Research Questions

### Study overview

This project examines how two major US sports betting brands, DraftKings and FanDuel, present gambling-related messages on their official YouTube channels. Rather than trying to estimate customer conversions, I will study the *content* of the messages: how betting promotions are combined with sports commentary or entertainment, what actions viewers are encouraged to take, and whether responsible-gambling information is visible. This is a descriptive content analysis, not an experiment about whether advertising changes behavior.

### Research questions

**RQ1.** What promotional techniques appear in the sampled videos (for example, bonuses, betting odds, urgency, direct calls to action, humor, or sports personalities)?

**RQ2.** How often are betting-related promotions embedded in sports analysis or entertainment rather than presented as clearly separated advertising?

**RQ3.** How frequently do sampled videos include visible or spoken responsible-gambling messages, age restrictions, or risk disclosures?

I expect to find a mixture of explicit promotions and entertainment-based marketing, but I will treat that as a tentative expectation rather than a result. The unit of analysis will be **one publicly available YouTube video**. My comparisons will be descriptive, not causal, and will not claim to measure audience demographics, sign-ups, or betting activity.


## A5 — Background & Related Work

Sports betting marketing is not limited to traditional advertisements. A systematic review by **Killick and Griffiths (2022)** examined 15 content-analysis studies of sports betting advertising and found recurring themes including humor and friendship. The review also noted that many studies did not report inter-coder reliability. This suggests that my project should use explicit coding definitions and check coding consistency rather than relying only on subjective impressions.

**Singer, Wöhr, and Otterbach (2024)** reviewed gambling operators' advertising strategies on social media and their effects. This work establishes social platforms as a relevant setting for investigating how operators present gambling products. My study narrows that broader topic to the messages used by two sports betting brands on YouTube, without claiming that content alone proves an effect on viewers.

**Rossi and Nairn (2024)** studied the recognizability of gambling content marketing and highlighted how advertising can blend with apparently ordinary social media content. This directly informs my second research question: I will distinguish videos that explicitly advertise betting from those that integrate promotional messages into sports entertainment.

Together, these studies motivate a small, transparent content analysis. They also point to an ethical concern: gambling marketing can be difficult to distinguish from ordinary sports content. My project adds a focused YouTube comparison, but a sample of company-published videos cannot establish what viewers noticed or how they behaved.

## A5 — Methodology

### 1. Sampling and data access

I will identify the **official** DraftKings and FanDuel YouTube channels and record their channel URLs and IDs before collecting data. My target sample is **60 public videos, 30 per company**, published during the six months ending November 15, 2026. Within each channel, I will list eligible videos in reverse chronological order and select the first 30, documenting any exclusions. I will exclude duplicates, inaccessible/deleted videos, and full-length livestream replays that cannot reasonably be coded within the project timeline; I will record how many items were excluded and why. If one channel has fewer than 30 eligible videos, I will report the actual sample and either extend the window consistently for both channels or reduce the target without silently replacing observations.

If I use the API, I will create a Google Cloud project, enable YouTube Data API v3, and obtain an API key without publishing it in GitHub. I will make a small test request first, check quota usage, and avoid expensive broad `search.list` calls where possible. Instead, I will retrieve each channel's uploads playlist ID using `channels.list`, enumerate its videos with `playlistItems.list`, and request only needed metadata using batched `videos.list` calls. I will document which calls I make, their quota costs, and whether the returned records are sufficient. If API setup or quota becomes a barrier, I will use a smaller manually documented sample of public video pages, subject to platform rules.

The primary evidence will be the **public video pages and the videos themselves**, reviewed manually. I will record video URL/ID, channel, publication date, title, length, and the relevant observed content. The YouTube Data API (https://developers.google.com/youtube/v3) is a possible aid for retrieving permitted metadata, **not a prerequisite** for coding videos. Before using API-derived information, I will check the YouTube API Services Terms and Developer Policies, including restrictions on derived content classifications and cross-channel performance comparisons. I will not scrape restricted information, collect viewer comments or identities, or publish a bulk copy of API data without confirming permission.

### 2. Coding scheme

I will create a short **codebook** before full coding. It will include a definition, yes/no rule, and concrete example for each indicator. I will watch each selected video rather than classifying it from titles and descriptions alone; for lengthy videos, I will document a consistent review protocol and flag any incomplete viewing. I will also examine video descriptions for betting links or promo codes, recording whether the promotion is in the video, description, or both.

I will create a spreadsheet with one row per video and the following manually observed variables:

| Variable | Operational definition |
|---|---|
| Brand and video link | DraftKings or FanDuel; public URL |
| Publication date and length | Date and duration shown on the video page |
| Primary format | Sports discussion, entertainment, explicit betting promotion, or other; choose the dominant format |
| Betting promotion present | Yes if the video includes a betting product, odds, offer, or instruction to place a bet |
| Bonus or incentive | Yes if a bonus, free bet, special offer, or similar incentive is mentioned |
| Direct call to action | Yes if viewers are explicitly asked to sign up, download, visit, or bet |
| Humor / celebrity or athlete | Separate yes/no indicators based on visible or spoken content |
| Embedded promotion | Yes if a betting message appears within otherwise sports-focused or entertainment content; a video can be both entertainment and promotional |
| Promo code / referral link | Separate yes/no fields for a visible or spoken promo code and a link/code in the description |
| Promotion location | In-video, description, or both; record timestamp or description excerpt |
| Responsible-gambling message | None, text-only, spoken, or both; note where it appears |
| Notes / evidence | Brief description and timestamp supporting ambiguous decisions |

A video may have several yes/no indicators, but only one primary format. Example codebook rules: **Entertainment** is the dominant format when the main purpose is humor, challenges, or light sports entertainment rather than detailed analysis; **embedded promotion = yes** when that same video includes betting odds, a sign-up prompt, or a promo code, even briefly. **Direct call to action = yes** only for an explicit instruction (for example, “sign up” or “use code X”), not simply showing a brand logo. **Responsible-gambling message = yes** only when a visible or spoken safer-gambling statement appears; a generic brand disclaimer alone will be recorded separately in notes. I will pilot-code 8–10 videos, revise unclear definitions, and recode those pilot videos using the final codebook. One week after initial coding, I will independently recode a randomly selected subset of 10 videos without looking at my earlier labels, calculate percentage agreement for the main binary categories, and document disagreements and any rule changes. This is a **within-coder consistency check**, not inter-coder reliability. If a classmate is available to code the same subset independently, I can separately report inter-coder agreement.

### 3. Analysis and presentation

I will use Python/pandas in Jupyter to count videos by primary format and calculate the proportion of sampled videos containing each promotional element, embedded promotion, and responsible-gambling messaging. I will present a sample-description table, a grouped bar chart of promotional techniques by brand, and a chart showing the presence of risk disclosures. Counts and percentages will always include denominators, and small differences will be treated cautiously rather than interpreted as statistically significant or causal.

I will include two or three brief, linked examples of videos to explain how I applied the coding definitions. I will avoid embedding personal data or extensive copyrighted material. The analysis will not rank the companies' business success, estimate conversions, or infer audience characteristics.

### 4. Feasibility, risks, and timeline

- **First:** Verify official channel identities, eligible video counts, API key setup, test-call quota, storage/retention rules, and whether videos can be manually reviewed within the available time.
- **Next:** Finalize the sample and codebook; pilot-code 8–10 videos.
- **Then:** Code the remaining sample, keep an audit trail of exclusions and ambiguous cases, and recode 10 videos one week later to check consistency; run descriptive analysis.
- **Finally:** Create charts, explain findings and limitations, and prepare the A7 report.

Potential obstacles include insufficient eligible videos, lengthy videos, changing availability, subjective classification, and platform policy constraints. If time becomes tight, I will reduce the sample transparently (for example, to 20 per company), prioritizing a consistent sampling rule and accurate coding over sample size. Since this is observational and limited to public company content, it cannot demonstrate advertising effectiveness, viewer exposure, or real-world gambling behavior.
